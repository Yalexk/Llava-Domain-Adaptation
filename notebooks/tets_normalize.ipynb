```python
from llava_peft.eval.metrics import normalize

# test birds
cases = [
    ("Kentucky Warbler", "kentucky warbler"),
    ("Kentucky warbler.", "kentucky warbler"),
    ("A photo of a mockingbird", "mockingbird"),
    ("Red-winged Blackbird", "red winged blackbird"),
    ("  the   Blue headed  vireo ", "blue headed vireo"),
    ("Acura TL Type-S 2008", "acura tl type s 2008"),
]

for text, want in cases:
    got = normalize(text)
    status = "PASS" if got == want else "FAIL"
    print(f"{status}  {text!r:32} -> {got!r}   (want {want!r})")
```

```python
from llava_peft.eval.metrics import exact_match, contains_match

# (pred, label, want_exact, want_contains)
match_cases = [
    ("Kentucky Warbler.", "kentucky warbler", True, True),
    ("This bird is a Kentucky warbler", "kentucky warbler", False, True),
    ("Mockingbird", "kentucky warbler", False, False),
]

for pred, label, want_exact, want_contains in match_cases:
    got = (exact_match(pred, label), contains_match(pred, label))
    status = "PASS" if got == (want_exact, want_contains) else "FAIL"
    print(f"{status}  {pred!r:36} vs {label!r:20} -> exact={got[0]}, contains={got[1]}")
```
