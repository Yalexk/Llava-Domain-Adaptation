```python
from llava_peft.utils import setup_env, get_device
from llava_peft.data import get_dataset
from llava_peft.model import load_model, generate

setup_env()
device = get_device()
model, processor = load_model(device=device)

spec = get_dataset("birds")
ds = spec.load("test", limit=8)
preds = generate(model, processor, ds[:4]["image"], spec.prompt)
for p, l in zip(preds, ds[:4]["label"]):
    print(f"pred: {p!r}   truth: {l!r}")
```
